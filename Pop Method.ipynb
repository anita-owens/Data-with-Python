```python
my_dict = {"a": 1, "b": 2, "c": 3}
my_dict
```

As you can see, pop() can be useful when you want to remove a specific item from a dictionary and do something with its value at the same time.

```python
value_of_b = my_dict.pop("b")  # removes the key "b" and returns its value, which is 2

print(my_dict)  # {"a": 1, "c": 3}
print(value_of_b)  # 2
```

```python
value_of_b = my_dict.pop("b", None)  # removes the key "b" and returns its value, which is 2

print(my_dict)  # {"a": 1, "c": 3}
print(value_of_b)  # 2
```
