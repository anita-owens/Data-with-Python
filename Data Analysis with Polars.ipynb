```python
data_import = (
    pl.scan_csv("datasets/marketing.csv")
   
)

marketing = data_import.collect()
```

```python
df1 = pl.LazyFrame(marketing)
```

```python
type(df1)
```

```python
ctx = pl.SQLContext(marketing=df1)

tables = ctx.execute("SHOW TABLES", eager=True)

print(tables)
```

```python
print(ctx.execute("SELECT * FROM marketing"))
```

```python
result = ctx.execute(
    """
        SELECT language_displayed, AVG(converted) as conversion_rate
        FROM marketing
        GROUP BY language_displayed
    """
)
print(result)
```
